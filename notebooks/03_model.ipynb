# FORESIGHT — Demand Forecasting Model

This notebook trains and evaluates the demand forecasting model.

Workflow:
1. Load processed data
2. Prepare time-series features
3. Build seasonal-naive baseline
4. Train Random Forest model
5. Evaluate using WAPE
6. Perform rolling-origin backtesting
7. Save forecast results

## Model Evaluation

Seasonal Naive WAPE: 11.97%

Random Forest WAPE: 10.99%

Improvement vs baseline: 8.23%

Rolling-origin WAPE: approximately 2.14%

## Conclusion

The Random Forest model achieved lower WAPE than the seasonal-naive
baseline on the evaluated test period.

The model was additionally evaluated using rolling-origin backtesting
to respect the chronological nature of the demand data.

The resulting forecasts are used as an input to the inventory risk
scoring workflow.
